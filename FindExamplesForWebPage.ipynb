```python
import json
import os
from IPython.display import display, Audio

def read_manifest(manifest_path):
    records = []
    with open(manifest_path, 'r') as f:
        all_lines = f.readlines()
        for line in all_lines:
            line = line.strip()
            record = json.loads(line)
            records.append(record)
    return records



technique_dirs = {
    "base_model": "/datap/misc/ICML_EVALS/base_checkpoint_evals/decodercontext_small_sp_ks3CorrectWithPrior_onlyphoneme_epoch161_Temp0.6_Topk80_Cfg_False_1.0_libri_unseen_test",
    "with_cfg": "/datap/misc/ICML_EVALS/cfg_scale_experiments/decodercontext_small_sp_ks3CorrectWithPrior_onlyphoneme_epoch161_Temp0.6_Topk80_Cfg_True_2.4_libri_unseen_test",
    "rpo" : "/datap/misc/ICML_EVALS/dpo_rpo_evals_paretofix/decodercontext_small_sp_ks3CorrectWithPrior_onlyphoneme_epoch161_RPO_epoch_1_Temp0.6_Topk80_Cfg_False_1.0_libri_unseen_test_shehzeen_phoneme",
    "rpo_cfg" : "/datap/misc/ICML_EVALS/dpo_rpo_evals_paretofix/decodercontext_small_sp_ks3CorrectWithPrior_onlyphoneme_epoch161_RPO_epoch_1_Temp0.6_Topk80_Cfg_False_1.0_libri_unseen_test_shehzeen_phoneme",
}

manifest_file = "/home/pneekhara/2023/SimpleT5NeMo/manifests/test_clean_withContextAudioPaths.json"
manifest_records = read_manifest(manifest_file)
filename_to_number = {}
for ridx, record in enumerate(manifest_records):
    filename_to_number[record['audio_filepath'].split("/")[-1]] = ridx
    
```

```python
def find_filewise_metrics(technique_dir):
    metrics_file = [f for f in os.listdir(technique_dir) if f.endswith("filewise_metrics_0.json")][0]
    metrics_file = os.path.join(technique_dir, metrics_file)
    with open(metrics_file) as f:
        metrics = json.load(f)
    
    filewise_metrics = {}
    for row in metrics:
        filename = row['gt_audio_filepath'].split("/")[-1]
        filewise_metrics[filename] = row
    
    return filewise_metrics
        
technique_metrics = {}
for technique in technique_dirs:
    technique_metrics[technique] = find_filewise_metrics(technique_dirs[technique])
```

```python
def is_good_candidate(technique_metrics, file):
    good_cer_contrast = (technique_metrics["base_model"][file]['cer'] > 0.05 and technique_metrics["base_model"][file]['cer'] < 0.25) and technique_metrics["with_cfg"][file]['cer'] == 0.0
    better_ssim = True
    better_cer = True
    for technique in ["with_cfg", "rpo"]:
        if technique_metrics[technique][file]['cer'] > technique_metrics["base_model"][file]['cer']:
            better_cer = False
        if technique_metrics[technique][file]['pred_context_ssim'] < technique_metrics["base_model"][file]['pred_context_ssim']:
            better_ssim = False
    if better_cer and better_ssim and good_cer_contrast:
        return True
    return False

num_good_candidates = 0
for file in technique_metrics["base_model"]:
    if is_good_candidate(technique_metrics, file):
        file_number = filename_to_number[file]
        print("Good candidate", file_number, num_good_candidates+1)
        print(technique_metrics["base_model"][file]['gt_audio_text'])
        for technique in technique_dirs:
            print(technique)
            print(technique_metrics[technique][file]['pred_text'])
            pred_file_path = os.path.join(technique_dirs[technique], "audio/predicted_audio_{}.wav".format(file_number))
            display(Audio(pred_file_path))
        print("-----------------")
        num_good_candidates += 1
            
        
```
